```python
# Matrix Chain Order using Dynamic Programming

# Number of matrices
n = int(input("Enter number of matrices: "))

# Enter dimensions
print("Enter matrix dimensions:")
p = []

for i in range(n + 1):
    p.append(int(input(f"p[{i}] = ")))

# Create DP table
dp = [[0] * n for _ in range(n)]

# Matrix Chain Order
for length in range(2, n + 1):
    for i in range(n - length + 1):
        j = i + length - 1
        dp[i][j] = float('inf')

        for k in range(i, j):
            cost = (
                dp[i][k]
                + dp[k + 1][j]
                + p[i] * p[k + 1] * p[j + 1]
            )

            dp[i][j] = min(dp[i][j], cost)

# Display result
print("\nMinimum number of scalar multiplications:")
print(dp[0][n - 1])

# Display DP table
print("\nDP Table:")
for row in dp:
    print(row)

# Complexity
print("\nTime Complexity: O(n^3)")
print("Space Complexity: O(n^2)")

```
